```python
quality_counts = wine['quality'].value_counts()
plt.figure(figsize=(8, 6))
plt.bar(quality_counts.indexUpdate, quality_counts, color='deeppink')
plt.title('Count Plot of Quality')
plt.xlabel('Quality')
plt.ylabel('Count')
plt.show()
```

```python
# Kernel density plots
sns.set_style("darkgrid")
numerical_columns = wine.select_dtypes(include=["int64", "float64"]).columns
plt.figure(figsize=(14, len(numerical_columns * 3)))
for idx, feature in enumerate(numerical_columns, 1):
    plt.subplot(len(numerical_columns), 2, idx)
    sns.histplot(wine[feature], kde=True, color='skyblue')
    plt.title(f"{feature} | Skewnewss: {round(wine[feature].skew(), 2)}")
    
plt.tight_layout()
plt.show()
```

```python
# Swarm plot to show outliers
plt.figure(figsize=(10, 8))

sns.swarmplot(data=wine, x="quality", y="alcohol", palette="viridis")
plt.title('Swarm Plot for Quanlity and Alcohol')
plt.xlabel('Quality')
plt.ylabel('Alcohol')
plt.show()
```

```python
sns.set_palette("Pastel1")
plt.figure(figsize=(10, 6))
sns.pairplot(wine)
plt.suptitle("Pair plot for DataFrame")
plt.show()
```

```python
wine['quality'] = wine['quality'].astype(str)
plt.figure(figsize=(10, 8))
sns.violinplot(x='quality', y='alcohol', data=wine, palette={'3':'lightcoral', '4':'lightblue', '5':'lightgreen', '6': 'gold', '7': 'lightskyblue', '8': 'lightpink'}, alpha=0.7)

plt.title('Violin Plot for Quality and Alcohol')
plt.xlabel('Quality')
plt.ylabel('Alcohol')
plt.show()
```

```python
sns.boxplot(x='quality', y='alcohol', data=wine)
```

```python
plt.figure(figsize=(15, 10))

sns.heatmap(wine.corr(), annot=True, fmt='.2f', cmap='Pastel2', linewidths=2)
plt.title('Correlation Heatmap')
plt.show()
```
